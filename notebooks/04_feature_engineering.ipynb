```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\hp5cd\Desktop\bank-churn-analytics\data\processed\customers1_updated.csv")
transactions = pd.read_csv(r"C:\Users\hp5cd\Desktop\bank-churn-analytics\data\synthetic\transactions.csv")
```

```python
df['balance_salary_ratio'] = (

    df['balance'] /

    (df['estimated_salary'] + 1)

)
# Does customer maintain high balance compared to salary?
```

```python
df['products_per_year'] = (

    df['products_number']

    /

    (df['tenure'] + 1)

)
# How actively customer adopted products
```

```python
total_txn = (

    transactions

    .groupby('customer_id')

    .size()

    .reset_index(name='total_transactions')

)
# Total Transactions 

df = df.merge(

    total_txn,

    on='customer_id',

    how='left'

)
```

```python
avg_txn = (

    transactions

    .groupby('customer_id')

    ['amount']

    .mean()

    .reset_index(name='avg_transaction_amount')

)

# Average Transaction Amount 

df = df.merge(

    avg_txn,

    on='customer_id',

    how='left'

)
```

```python
digital_txn = (

    transactions[
        transactions['channel']
        == 'Mobile App'
    ]

    .groupby('customer_id')

    .size()

    .reset_index(name='digital_transactions')

)

# Digital Usage Ratio

df = df.merge(

    digital_txn,

    on='customer_id',

    how='left'

)
```

```python
df['digital_transactions'] = (

    df['digital_transactions']

    .fillna(0)

)
```

```python
df['digital_txn_ratio'] = (

    df['digital_transactions']

    /

    (df['total_transactions'] + 1)

)
```

```python
df.to_csv(

    r"C:\Users\hp5cd\Desktop\bank-churn-analytics\data\feature_engineering\feature_engineered_data.csv",

    index=False

)
```
