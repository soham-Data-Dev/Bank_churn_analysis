```python
import pandas as pd 
customers = pd.read_csv(r'C:\Users\hp5cd\Desktop\bank-churn-analytics\data\raw\Churn_Modelling (1).csv')
```

```python
transactions = []

transaction_types = [
    'Salary',
    'ATM',
    'UPI',
    'Shopping',
    'Bill Payment'
]

merchant_categories = {
    'Salary': ['Company Payroll'],
    'ATM': ['Cash Withdrawal'],
    'UPI': ['Transfer', 'Food', 'Recharge'],
    'Shopping': ['Amazon', 'Flipkart', 'Myntra'],
    'Bill Payment': ['Electricity', 'Internet', 'Mobile']
}

channels = ['Mobile App', 'ATM Machine', 'Internet Banking', 'Branch']

transaction_id = 100000

for customer_id in customers['CustomerId']:

    # random number of transactions per customer
    num_transactions = random.randint(20, 300)

    for _ in range(num_transactions):

        txn_type = random.choice(transaction_types)

        merchant = random.choice(
            merchant_categories[txn_type]
        )

        # random transaction amount
        if txn_type == 'Salary':
            amount = random.randint(30000, 150000)

        elif txn_type == 'ATM':
            amount = random.randint(500, 10000)

        elif txn_type == 'UPI':
            amount = random.randint(50, 5000)

        elif txn_type == 'Shopping':
            amount = random.randint(200, 25000)

        else:
            amount = random.randint(300, 15000)

        # random date in last 2 years
        random_days = random.randint(0, 730)

        txn_date = datetime.now() - timedelta(days=random_days)

        transactions.append({
            'transaction_id': transaction_id,
            'customer_id': customer_id,
            'date': txn_date,
            'amount': amount,
            'transaction_type': txn_type,
            'merchant_category': merchant,
            'channel': random.choice(channels)
        })

        transaction_id += 1
```

```python
transactions_df = pd.DataFrame(transactions)
```
