```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\hp5cd\Desktop\bank-churn-analytics\data\processed\customers1_updated.csv")
```

1] Churn Distribution 
   Measures overall churn severity.

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x='churn',
    data=df
)

plt.title("Churn Distribution")

plt.savefig(
    r"C:\Users\hp5cd\Desktop\bank-churn-analytics\visuals\churn_distribution.png"
)

plt.show()
```

2] Churn VS Age 
   Checks whether older customers churn more frequently.

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='churn',
    y='age',
    data=df
)

plt.title("Churn vs Age")

plt.savefig(
    r"C:\Users\hp5cd\Desktop\bank-churn-analytics\visuals\churn_vs_age.png"
)

plt.show()
```

3] Churn vs Country
   Identifies risky geographic regions.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='country',
    hue='churn',
    data=df
)

plt.title("Churn by Country")

plt.savefig(
    r"C:\Users\hp5cd\Desktop\bank-churn-analytics\visuals\churn_vs_country.png"
)

plt.show()
```

4]Churn vs Balance
Detects whether high-value customers are leaving.

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='churn',
    y='balance',
    data=df
)

plt.title("Churn vs Balance")

plt.savefig(
     r"C:\Users\hp5cd\Desktop\bank-churn-analytics\visuals\churn_vs_balance.png"
)

plt.show()
```

5] Churn VS Product
Measures customer stickiness through product ownership.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='products_number',
    hue='churn',
    data=df
)

plt.title("Churn vs Products")

plt.savefig(
    r"C:\Users\hp5cd\Desktop\bank-churn-analytics\visuals\churn_vs_products.png"
)

plt.show()


```

6] Churn VS Active Members
Determines whether inactive customers are more likely to churn.plt.figure(figsize=(8,5))

```python
sns.countplot(
    x='active_member',
    hue='churn',
    data=df
)

plt.title("Churn vs Active Membership")

plt.savefig(
    r"C:\Users\hp5cd\Desktop\bank-churn-analytics\visuals\churn_vs_active_member.png"
)

plt.show()
```

### Insight

Inactive customers show significantly higher churn rates, suggesting customer engagement is strongly linked with retention.
